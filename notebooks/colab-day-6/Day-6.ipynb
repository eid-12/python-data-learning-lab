## **TASK-1**

```python
try:
  name : str = input("Enter your name: ")
  age : int = int(input("Enter your age: "))

  if age < 0 or age > 116:
     raise Exception("invalid age")

  favourite_color : str = input("Enter your favourite color: ")

  print(f"\nName: {name}\nage: {age}\nFavourite color: {favourite_color}.")
except ValueError as e:
  print(f"Wrong Type inputted: {e}")
```

## **TASK-2**

```python
import random

list_of_nums = [random.randint(20,1500) for x in range(15)]
sum = 0
for x in list_of_nums:
  sum += x
average = sum/len(list_of_nums)
print(list_of_nums)
print(f"Sum: {sum}\nAverage: {average}")
```

## **TASK-3**

```python
list_of_nums = [random.randint(10,20) for x in range(15)]
list_of_bools = []
for x in list_of_nums:
  if x > 0:
    list_of_bools.append(True)
  else:
    list_of_bools.append(False)
print(f"Array of numbers: {list_of_nums}")
print(f"Array of boolean values: {list_of_bools}")
```

## **TASK-7**

```python
def index(arr : list, var):
  for x in range(len(arr)):
    if arr[x] == var:
      return x
  return "Value does not exist inside of list"
l = [10,20,30,40,50]
print(index(arr=l,var=30))
```
