```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv('happyscore_income.csv')
df.head()
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
df.describe()
```

```python
x=df[['avg_income']]
y=df[['happyScore']]
```

```python

x_traing,x_test,y_traing,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
reg=linear_model.LinearRegression()
reg.fit(x_traing,y_traing)
```

```python

x_pred=reg.predict(x_test)
x_pred=x_pred.reshape(-1,1)
r2_score(x_test,x_pred)
```

```python
plt.xlabel('income')
plt.ylabel('happyScore')
plt.scatter(x_test, y_test, color='red', marker='+' )
plt.plot(x_test, x_pred, color='blue')
```
