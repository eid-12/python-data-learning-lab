# **1. Python Programming**

```python
import pandas as pd
import numpy as np

data = pd.read_csv("ecommerce_sales.csv")

data.head(10)
```

# **2.Data Cleaning**

```python

print("Before removing duplicates and null values:", data.shape)
data.dropna(inplace=True)
data.drop_duplicates(inplace=True)
print("After removing duplicates and null values:", data.shape)
```

```python
print("Remove Negative values :")
data = data[(data['Quantity'] >= 0) & (data['UnitPrice'] >= 0)]

data.describe()
```

```python
data['TotalPrice'] = data['Quantity'] * data['UnitPrice']
data.head()
```

# **3.EDA**

```python
print("Basic Statistics:")
data.describe()
```

```python
print("Top 10 Selling Products by Quantity:")
top_products=data.groupby(['Description'])['Quantity'].sum().sort_values(ascending=False).head(10)
```

```python
print("Revenue Analysis:")
total_revenue = data['TotalPrice'].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```
