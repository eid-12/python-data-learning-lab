```python
print("Loading Fashion MNIST dataset...")
(x, y), (x_test, y_test) = fashion_mnist.load_data()
x = x.reshape(-1, 28*28)
x_test = x_test.reshape(-1, 28*28)
# Normalize pixel values to [0,1]
x = x / 255.0
x_test = x_test / 255.0
```

```python
print ("splitting data:")
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.1, random_state=42)
```

```python
print("Implementing SVM.")
X_small = x_train[:10000]
y_small = y_train[:10000]
```

```python
print("Visualizing...")
list =['linear', 'poly', 'rbf']
t = ''
testccuracy=0
for i in list:
    models = SVC(kernel=i, random_state=42)
    models.fit(X_small, y_small)
    val_accuracy = models.score(x_val, y_val)
    print(f" {i}:Validation set accuracy:", val_accuracy)
    test_accuracy = models.score(x_test, y_test)
    if test_accuracy > testccuracy:
        t=i
        testccuracy=test_accuracy

    print("Test set accuracy:", test_accuracy)
    print("\n")
```

```python
model = SVC(kernel= t , random_state=42)
model.fit(X_small, y_small)
val_accuracy = model.score(x_val, y_val)
print("Validation set accuracy:", val_accuracy)
```

```python
test_accuracy = model.score(x_test, y_test)
print("Test set accuracy:", test_accuracy)
```

```python
y_pred = model.predict(x_test)
```

```python
print(f"Evaluating by accuracy:{t}")
accuracy = accuracy_score(y_test, y_pred)
print(f'\nAccuracy: {accuracy * 100:.2f}%')

conf_matrix = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(conf_matrix)

fig, ax = plt.subplots(figsize=(8, 6))
cax = ax.matshow(conf_matrix, cmap='coolwarm')
fig.colorbar(cax)

ax.set_title('Confusion Matrix Heatmap')
ax.set_xlabel('Predicted Labels')
ax.set_ylabel('True Labels')

ax.set_xticks(np.arange(len(np.unique(y_test))))
ax.set_yticks(np.arange(len(np.unique(y_test))))

ax.xaxis.set_label_position('top')
ax.xaxis.tick_top()

plt.show()
```
