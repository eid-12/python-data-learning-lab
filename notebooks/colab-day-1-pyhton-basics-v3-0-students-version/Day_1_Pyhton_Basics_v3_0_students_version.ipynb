**Comparision Operators:**

1- Equal to (==): This operator checks whether the values on both sides are equal. If they are, the result is True.

2- Not equal to (!=): This operator checks whether the values on both sides are not equal. If they aren't, the result is True.

3- Greater than (>): This operator checks if the value on the left is greater than the value on the right. If it is, the result is True.

4- Less than (<): This operator checks if the value on the left is less than the value on the right. If it is, the result is True.

5- Greater than or equal to (>=): This operator checks if the value on the left is greater than or equal to the value on the right. If it is, the result is True.

6- Less than or equal to (<=): This operator checks if the value on the left is less than or equal to the value on the right. If it is, the result is True.

**Logical operators in Python are used to combine the results of two or more Boolean expressions into a single expression. These operators evaluate expressions to return True or False based on the condition(s).**

Python includes three primary logical operators:

**and:**

Usage: Evaluates two conditions and returns True if both conditions are true, otherwise returns False.
Example: (5 > 3) and (6 > 4) evaluates to True because both comparative expressions are true.

**or:**

Usage: Evaluates two conditions and returns True if at least one of the conditions is true. If both conditions are false, it returns False.
Example: (5 > 10) or (4 > 2) evaluates to True because one of the conditions (4 > 2) is true.

**not:**

Usage: Negates the result of the condition that follows it. If the condition is True, not converts it to False, and vice versa.
Example: not(5 > 10) evaluates to True because 5 > 10 is false, and not inverts it.

3- Case Sensitivity: Python treats identifiers as case-sensitive, meaning variable, Variable, and VARIABLE are seen as distinct.

4-Reserved Words Prohibition: Python’s reserved words or keywords, like if, for, return, class, cannot be used as identifiers.

## Task 5: Prime Numbers Test

````python
while True:
    n = int(input("Please enter a number or Exit Enter 0 "))
    if n == 0:
        print("Thank you for using the program")
        break
    elif n >2:
        for i in range(2, n):
            if n % i == 0:
                print(f"{n} is not a prime number")
                break
        else:
            print(f"{n} is a prime number")
    elif n == 2:
        print(f"{n} is a prime number")
    else:
        print(f"{n} is not a prime number")
````

Here’s a detailed explanation of how user-defined functions work in Python, their benefits, and how to use them effectively.

Basic Usage of User-Defined Functions
Creating a Function:

Definition: Functions are defined using the def keyword followed by a unique function name and parentheses which may include parameters. The code block within every function starts with a colon and is indented.



```

    # code block

```


Calling a Function:

After defining a function, you can call it by using its name followed by parentheses. If the function requires arguments, you provide them inside these parentheses.



```

```

Parameters and Arguments:

Parameters are variables listed in the function definition and act as placeholders for values that are passed to the function when it is called.
Arguments are the actual values you pass to the function when you call it.
Functions can have any number of parameters or none at all.



Positional Parameters:
